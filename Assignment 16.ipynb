```python
import sqlite3
import pandas as pd

# Connect to your Chinook database
conn = sqlite3.connect('chinook.db')

# Define your master query
master_query = """
SELECT analysis_section, item_name, primary_metric, secondary_metric FROM (
    SELECT '1. Top Genre by Tracks' AS analysis_section, g.name AS item_name, COUNT(t.track_id) || ' tracks' AS primary_metric, NULL AS secondary_metric
    FROM genre g JOIN track t ON g.genre_id = t.genre_id GROUP BY g.genre_id, g.name ORDER BY COUNT(t.track_id) DESC LIMIT 3
) t1

UNION ALL

SELECT analysis_section, item_name, primary_metric, secondary_metric FROM (
    SELECT '3. Top Customer by Spending' AS analysis_section, c.first_name || ' ' || c.last_name AS item_name, '$' || ROUND(SUM(i.total), 2) AS primary_metric, 'ID: ' || c.customer_id AS secondary_metric
    FROM customer c JOIN invoice i ON c.customer_id = i.customer_id GROUP BY c.customer_id ORDER BY SUM(i.total) DESC LIMIT 3
) t2

UNION ALL

SELECT analysis_section, item_name, primary_metric, secondary_metric FROM (
    SELECT '4. Top Artist by Revenue' AS analysis_section, ar.name AS item_name, '$' || ROUND(SUM(il.unit_price * il.quantity), 2) AS primary_metric, NULL AS secondary_metric
    FROM artist ar JOIN album al ON ar.artist_id = al.artist_id JOIN track t ON al.album_id = t.album_id JOIN invoice_line il ON t.track_id = il.track_id GROUP BY ar.artist_id, ar.name ORDER BY SUM(il.unit_price * il.quantity) DESC LIMIT 3
) t3

UNION ALL

SELECT analysis_section, item_name, primary_metric, secondary_metric FROM (
    SELECT '5. Top Country by Avg Invoice' AS analysis_section, billing_country AS item_name, '$' || ROUND(AVG(total), 2) AS primary_metric, COUNT(invoice_id) || ' invoices' AS secondary_metric
    FROM invoice GROUP BY billing_country ORDER BY AVG(total) DESC LIMIT 3
) t4

UNION ALL

SELECT analysis_section, item_name, primary_metric, secondary_metric FROM (
    SELECT '6. Top Media Type' AS analysis_section, m.name AS item_name, COUNT(t.track_id) || ' tracks' AS primary_metric, NULL AS secondary_metric
    FROM media_type m JOIN track t ON m.media_type_id = t.media_type_id GROUP BY m.media_type_id, m.name ORDER BY COUNT(t.track_id) DESC LIMIT 3
) t5

UNION ALL

SELECT analysis_section, item_name, primary_metric, secondary_metric FROM (
    SELECT '8. Longest Track' AS analysis_section, t.name AS item_name, ROUND(t.milliseconds / 1000.0, 2) || ' sec' AS primary_metric, ar.name AS secondary_metric
    FROM track t JOIN album al ON t.album_id = al.album_id JOIN artist ar ON al.artist_id = ar.artist_id ORDER BY t.milliseconds DESC LIMIT 3
) t6;
"""

# Execute the query and display the DataFrame
df = pd.read_sql(master_query, conn)
df
```
